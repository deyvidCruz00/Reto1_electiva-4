El diccionario, define una relación uno a uno entre una clave y su valor.

Los diccionarios en Python, al igual que las listas y las tuplas, permiten almacenar diferentes tipos de datos: Strings, enteros, flotantes, booleanos, tuplas, listas e inclusive otros diccionarios.

Los diccionarios son **mutables**, es decir, es posible modificar su longitud, podemos agregar o quitar elementos de él; de igual forma todos los valores almacenados en el diccionario pueden ser modificados.

# copy()

Este método devuelve una copia del diccionario:

```python
versiones = dict(python=2.7, zope=2.13, plone=5.1)
otro_versiones = versiones.copy()
print(versiones == otro_versiones)
print(type(otro_versiones))
```

# fromkeys()

Este método crea un nuevo diccionario con claves a partir de un tipo de dato secuencia. El valor de value por defecto es el tipo None.

```python
secuencia = ('python', 'zope', 'plone')
versiones = dict.fromkeys(secuencia)
print("Nuevo Diccionario : %s" %  str(versiones))
print (versiones)
```

En el ejemplo anterior inicializa los valores de cada clave a None, también se puede inicializar con un valor común por defecto para cada clave:

```python
versiones = dict.fromkeys(secuencia, 0.1)
print("Nuevo Diccionario : %s" %  str(versiones))
```

**ENUMERATE**
Enumerate es una función incorporada de Python. Permite recorrer una estructura de datos y asignar un contador automático. La estructura es: for counter, value in enumerate(some_list): print(counter, value)

```python
my_list = ['apple', 'banana', 'grapes', 'pear']
for c, value in enumerate(my_list, 1):
    print(c, value)
```

```python
my_list = ['apple', 'banana', 'grapes', 'pear']
counter_list = list(enumerate(my_list, 1))
print(counter_list)
```

```python
some_list = ['foo', 'bar', 'baz']
mapping = dict((k, v) for v, k in enumerate(some_list))
print(mapping)
```

Construcción de Diccionarios a partir de dos listas  utilizadon **ZIP** Function.

**What is zip method in Python?**

The zip() function returns a zip object, which is an iterator of collections, where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

```python
key_list=['id','name','salary']
value_list=[10,'Hugo',31.34]
empleado = {}
for key, value in zip(key_list, value_list):
    empleado[key] = value
print(empleado)
value = empleado.get('name')
print(value)
```

```python
words = ['apple', 'bat', 'bar', 'atom', 'book', 'cat']
palabras_por_letra = {}
for palabra in words:
    letra = palabra[0]
    palabras_por_letra.setdefault(letra, []).append(palabra)

print(palabras_por_letra)
```

```python
def contar_vocales_y_consonantes(cadena):
    resultado = {vocal: 0 for vocal in 'aeiou'}
    consonantes = 0
    for caracter in cadena.lower():
        if caracter in resultado:
            resultado[caracter] += 1
        elif caracter.isalpha():
            consonantes += 1
    resultado['Consonantes'] = consonantes
    return resultado

print(contar_vocales_y_consonantes('Gestion de Datos'))
```

```python
datos = ('id;nombre;correo;movil;salario\n'
         '3412;Pepe Perez;pepeperez@yahoo.com;300281234;150000\n'
         '45342;Maria Melo;mariamelo@yahoo.com;315434223;300000\n'
         '5673321;Fernando Jimenez;ferjim@gmail.com;312342234;230000\n'
         '4545231;Carlos Cardenas;carloscardenas@hotmail.com;3156754323;345000')

lineas = datos.splitlines()
encabezados = lineas[0].split(';')
empleados = {}
for linea in lineas[1:]:
    valores = linea.split(';')
    registro = dict(zip(encabezados, valores))
    identificacion = registro.pop('id')
    empleados[identificacion] = registro

print(empleados)
```
