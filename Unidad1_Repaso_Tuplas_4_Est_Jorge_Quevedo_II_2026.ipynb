# Tuplas

**Las tuplas son inmutables**

Una tupla es una secuencia de valores muy parecida a una lista. Los valores almacenados en una tupla pueden ser de cualquier tipo, y están indexados por números enteros. Las tuplas también son comparables y hashables (Un objeto es hashable si tiene un valor hash que nunca cambia durante su vida útil), por lo que se pueden ordenar y utilizarlas como valores clave en los diccionarios.

Sintácticamente, una tupla es una lista de valores separada por comas.

```python
t = 'a', 'b', 'c', 'd', 'e'
```

```python
print(type(t))
print(t)
```

```python
t=('a', 'b', 'c', 'd', 'e')
print(type(t))
print(t)
```

Aunque no es necesario, es común encerrar las tuplas entre paréntesis **()** para identificar las tuplas cuando se escribe código en Python:

```python
t1 = ('esto es una cadena', 123, 'c', 20.8, 'e',(32.2,32),[89,"Hola"])
print(t1)
t1[0]
```

```python
#  [] ---> Operador bracket
t1[0:3]
```

```python
print(t1[5][0])
print(t1[6][1][3])
```

Para crear una tupla con un solo elemento, hay que incluir la coma final:

```python
t1 = ('a',)
type(t1)
```

```python
print(t1)
```

Sin la coma Python trata ('a') como una expresión con una cadena entre paréntesis que evalúa a una cadena:

```python
t2 = ('a')
type(t2)
```

Otra forma de construir una tupla es la tupla de función incorporada. Sin ningún argumento, crea una tupla vacía:

```python
t = tuple()
print(type(t))
```

```python
print(t)
```

Si el argumento es una secuencia (cadena, lista o tupla), el resultado de la llamada a la tupla es una tupla con los elementos de la secuencia:

```python
t = tuple('Gestión de Datos')
print(t)
```

La mayoría de los operadores de listas también trabajan con tuplas. El operador bracket **[]** indexa un elemento:

```python
print(t[1:3])
```

Pero si intenta modificar uno de los elementos de la tupla, obtiene un error:

```python
t[0] = 'A'
```

No se pueden modificar los elementos de una tupla, pero se puede reemplazar una tupla por otra:

```python
t = ('A',) + t[1:]
print(t)
```

**Métodos**

**count()**
Este método recibe un elemento como argumento, y cuenta la cantidad de veces que aparece en la tupla.

```python
t.count('t')
```

**index()**

Comparte el mismo método index() del tipo lista. Este método recibe un elemento como argumento, y devuelve el índice de su primera aparición en la tupla.

```python
t.index('e')
```
