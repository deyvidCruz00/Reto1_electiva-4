**String**

Este cuadernillo presenta algunas técnicas de manipulación de cadenas de caracteres (o strings) en Python. Saber cómo manipular cadenas de caracteres es fundamental para las tareas de procesamiento de texto.

Un string es una secuencia de caracteres. Puede acceder a los caracteres de uno en uno con el **Operador Brackets [ ]**:



```python
asignatura="Gestion de Datos"
type(asignatura)
```

```python
letra=asignatura[4]
print(letra)
```

La anterior sentencia extrae el carácter en la posición 4 del índice de la variable **asignatura** y lo asigna a la variable **letra**. La expresión entre brackets se llama índice. El índice indica el carácter de la secuencia que se desea (de ahí el nombre).

El string o cadena es un arreglo unidimensional, es decir, que en este caso el índice de la primera letra corresponde a cero (0)  y así se sigue enumerando el índice para las demás letras del arreglo.

```python
letra=asignatura[11]
print(letra)
```

Para obtener la longitud de un string usando la función (implementada en las librerias de python) **len()** la cual devuelve el número de caracteres de una cadena:

```python
cad="La casa se pinocho"
print(len(cad))
```

Obtener la última letra de un string

```python
longitud = len(asignatura)
print(longitud)
```

```python
# Asignar la última letra de la dadena
ultima = asignatura[longitud]
print(ultima)
```

La razón del IndexError es que no hay ninguna caracter con el índice 22. El indice se inicia a contar desde cero, todos los carácteres están numerados del 0 hasta el máximo menos 1. Para obtener el último carácter, hay que restarle 1 a la longitud de la cadena:

```python
ultima = asignatura[longitud-1]
```

```python
print(ultima)
```

Alternativamente, puede usar índices negativos, que cuentan hacia atrás desde el final de la cadena. La expresión asignatura[-1] da la última letra, asignatura[-2] da la el penúltimo, y así sucesivamente.

```python
print(asignatura[-2])
```

**Seleccionar un fragmento de un String**

Un segmento de un string se llama un trozo. Seleccionar una rebanada es similar a seleccionar más de un caracter:

```python
s='Gestión de Datos'
print(s[0:5])
```

```python
print(s[8:11])
```

El operador devuelve la parte de la cadena del "n-ésimo" carácter "m-ésimo" al carácter, incluyendo el primero pero excluyendo el último [n,m). Si se omite el primer índice (antes de los dos puntos), el trozo comienza al principio de la cadena. Si se omite el segundo índice, la rebanada va al final de la cadena:

```python
s[:3]
```

```python
s[3:]
```

Si el primer índice es mayor o igual que el segundo el resultado es una cadena vacía, representada por dos comillas:

```python
s[3:3]
```

**Los string son inmutables.**

```python
#
fac='Ingeniería'
fac[0] = 'J'
```

```python
fac='Ingeniería'
nueva_fac = 'J' + fac[1:]
print(nueva_fac)
```

El operador **in**

```python
'b' in fac
```

El método **format()** de string
Se pueden definir variables y poner corchetes {} en el string donde van a ir los números o caracteres. Entre paréntesis pasamos los valores. Unos ejemplos sencillos.

```python
# saca "El valor es 69
print ("El valor es {}".format(69))

# saca "El valor es 12.3456
print ("El valor es {}".format(12.3456))

# Tres conjuntos {}, el primero para el primer parámetro de format(),
# el segundo para el segundo
# y así sucesivamente.
# saca "Los valores son 1, 2 y 3"
print ("Los valores son {}, {} y {}".format(1,2,3))

# Entre las llaves podemos poner la posición del parámetro.
# {2} es el tercer parámetro de format()
# {0} es el primer parámetro de format.
# saca "Los valores son 3, 2 y 1"
print ("Los valores son {2}, {1} y {0}".format(1,2,3))
```
